## 1 -Teste dos Sinais (Amostras Pareadas)

Um fisioterapeuta quer saber se um novo exercício respiratório reduz a frequência respiratória dos pacientes.
Foram medidos os valores antes e depois em 8 pessoas:

| **Paciente** | **Antes** | **Depois** |
| ------------ | --------- | ---------- |
| 1            | 22        | 20         |
| 2            | 24        | 23         |
| 3            | 19        | 18         |
| 4            | 26        | 25         |
| 5            | 21        | 22         |
| 6            | 23        | 21         |
| 7            | 25        | 25         |
| 8            | 20        | 19         |

```python
import numpy as np

# Declare a tabela como numpy
paciente = np.array([1, 2, 3, 4, 5, 6, 7, 8])
antes = np.array([22, 24, 19, 26, 21, 23, 25, 20])
depois = np.array([20, 23, 18, 25, 22, 21, 25, 19])
tabela = np.array([paciente, antes, depois])
tabela
```

##### 1.1 - Quais são as hipóteses do Teste dos Sinais?

**Resposta:**
- **H0**: Hipótese de nulidade, que nesse caso representa que não há uma queda significativa na frequência respiratória dos pacientes antes e depois do novo exercício, o que atestaria para a inefecácia do exercício para esse grupo.
- **H1**: Hipótese alternativa, que nesse caso representa que há uma queda significativa na frequência respiratória dos pacientes antes e depois do novo exercício, o que atestaria para a eficácia do exercício para esse grupo.

##### 1.2 - Identifique diferenças positivas, negativas e empates

```python
diferenca = depois - antes
diferenca
```

##### 1.3 - Gere um gráfico de barras mostrando as diferenças

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.barplot(x=paciente, y=diferenca)
plt.title('Diferença na frequência respiratória dos pacientes antes e depois do novo exercício')
plt.xlabel('Paciente')
plt.ylabel('Diferença na frequência respiratória')
plt.tight_layout()
plt.show()
```

##### 1.4 - Calcule o p-valor

```python
diferencas_sem_empates = diferenca[diferenca != 0]

n_positivos = np.sum(diferencas_sem_empates > 0)
n_total = len(diferencas_sem_empates)

from scipy.stats import binom

p_valor = binom.cdf(n_positivos, n_total, 0.5)

print(f"Valor de P pelo Teste dos Sinais: {p_valor:.4f}")
```

##### 1.5 - Conclusão

**Resposta:** Baseado no valor de p(6,25%), decide-se por aceitar H0, o que indica que não evidência estática suficiente que ateste para a eficácia do exercício respiratório que reduz a frequência respiratória para esse grupo.
